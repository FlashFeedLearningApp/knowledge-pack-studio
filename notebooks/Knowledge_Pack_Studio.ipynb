Turn an idea into an evidence-led FlashFeed knowledge pack. This notebook launches a guided visual application; you do not need to edit Python.

1. In Colab, add `OPENAI_API_KEY_FF_KP` under **Secrets** if you want live OpenAI research and generation.

> API usage is billed to the key you provide. The application never writes the key into artifacts or exports.

```python
# @title Install the pinned Knowledge Pack Studio package
from pathlib import Path
import os

IN_COLAB = Path('/content').is_dir() or 'COLAB_RELEASE_TAG' in os.environ
INSTALL_SPEC = 'git+https://github.com/FlashFeedLearningApp/knowledge-pack-studio.git@main' if IN_COLAB else '-e .[dev]'
%pip install -q $INSTALL_SPEC
```

```python
# @title Connect your OpenAI credential and Studio password
COLAB_SECRET_NAME = 'OPENAI_API_KEY_FF_KP'
COLAB_PASSWORD_SECRET_NAME = 'FF_KP_STUDIO_PASSWORD'
api_key = None
studio_password = None
if IN_COLAB:
    try:
        from google.colab import userdata
        api_key = userdata.get(COLAB_SECRET_NAME)
        studio_password = userdata.get(COLAB_PASSWORD_SECRET_NAME)
    except Exception:
        api_key = None
        studio_password = None

print(f'{COLAB_SECRET_NAME} connected.' if api_key else f'{COLAB_SECRET_NAME} not found. Start in mock mode or enter a key in the Studio UI.')
print(f'{COLAB_PASSWORD_SECRET_NAME} connected.' if studio_password else f'{COLAB_PASSWORD_SECRET_NAME} is required for a protected Colab launch.')
```

```python
# @title Launch Knowledge Pack Studio
from knowledge_pack_studio import launch

if IN_COLAB and not studio_password:
    raise RuntimeError(f'Add {COLAB_PASSWORD_SECRET_NAME} to Colab Secrets before launching.')

auth = ('ff-kp-author', studio_password) if studio_password else None
launch(api_key=api_key, auth=auth)
```
