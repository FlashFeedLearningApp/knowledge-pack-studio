Turn an idea into an evidence-led FlashFeed knowledge pack. This notebook launches a guided visual application; you do not need to edit Python.

> The Studio stays inside Colab's authenticated private runtime proxy; it does not create a public Gradio link. API usage is billed to the key you provide, and the application never writes the key into artifacts or exports.

> The final **Launch Knowledge Pack Studio** cell intentionally stays running so server errors and logs appear directly beneath it. There is no separate error-display setting to find. If you previously launched an unstyled page, choose **Runtime → Restart session**, then **Run all** to replace the old server.

```python
# @title Install the pinned Knowledge Pack Studio package
import os
import subprocess
import sys
from pathlib import Path

IN_COLAB = Path("/content").is_dir() or "COLAB_RELEASE_TAG" in os.environ
if any(
    name == "knowledge_pack_studio" or name.startswith("knowledge_pack_studio.")
    for name in sys.modules
):
    raise RuntimeError(
        "An older Studio build is already loaded in this runtime. Choose Runtime → Restart "
        "session, then Run all so Python cannot reuse cached code."
    )

if IN_COLAB:
    runtime_dependencies = [
        "gradio==6.22.0",
        "jsonschema==4.26.0",
        "openai==2.53.0",
        "pydantic==2.13.4",
    ]
    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "--upgrade",
            "--no-cache-dir",
            *runtime_dependencies,
        ],
        check=True,
    )
    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "--force-reinstall",
            "--no-cache-dir",
            "--no-deps",
            "git+https://github.com/FlashFeedLearningApp/knowledge-pack-studio.git@main",
        ],
        check=True,
    )
else:
    subprocess.run(
        [sys.executable, "-m", "pip", "install", "-q", "-e", ".[dev]"],
        check=True,
    )

print(
    "Fresh Knowledge Pack Studio build installed. Exact build details will appear in the launch cell."
)
```

```python
# @title Connect your OpenAI Colab Secret
COLAB_SECRET_NAME = "OPENAI_API_KEY_FF_KP"
api_key = None
if IN_COLAB:
    try:
        from google.colab import userdata

        api_key = userdata.get(COLAB_SECRET_NAME)
    except Exception:
        api_key = None

print(
    f"{COLAB_SECRET_NAME} connected."
    if api_key
    else f"{COLAB_SECRET_NAME} not found. Live mode is unavailable; start in mock mode."
)
```

```python
# @title Launch Knowledge Pack Studio
from knowledge_pack_studio import launch

launch(api_key=api_key, share=False, debug=True)
```
